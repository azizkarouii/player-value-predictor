## 1. Chargement de tous les datasets

```python
# Définir les chemins vers tous les datasets
datasets_paths = {
    'player_injuries': 'player_injuries/player_injuries.csv',
    'player_latest_market_value': 'player_latest_market_value/player_latest_market_value.csv',
    'player_market_value': 'player_market_value/player_market_value.csv',
    'player_national_performances': 'player_national_performances/player_national_performances.csv',
    'player_performances': 'player_performances/player_performances.csv',
    'player_profiles': 'player_profiles/player_profiles.csv',
    'player_teammates_played_with': 'player_teammates_played_with/player_teammates_played_with.csv',
    'team_children': 'team_children/team_children.csv',
    'team_competitions_seasons': 'team_competitions_seasons/team_competitions_seasons.csv',
    'team_details': 'team_details/team_details.csv',
    'transfer_history': 'transfer_history/transfer_history.csv'
}

# Charger tous les datasets
datasets = {}
print("Chargement des datasets :")

for name, path in datasets_paths.items():
    try:
        datasets[name] = pd.read_csv(path)
        print(f"  + {name}: {len(datasets[name]):,} lignes, {len(datasets[name].columns)} colonnes")
    except FileNotFoundError:
        print(f"  - {name}: Fichier non trouvé - {path}")
    except Exception as e:
        print(f"  - {name}: Erreur lors du chargement - {e}")

print(f"\nDatasets chargés avec succès : {len(datasets)}/{len(datasets_paths)}")
```

## 2. Player Injuries - Analyse des blessures des joueurs

```python
if 'player_injuries' in datasets:
    df = datasets['player_injuries']
    print("PLAYER INJURIES - Blessures des joueurs")
    print("=" * 50)
    
    print(f"Dimension du dataset : {df.shape}")
    print(f"Nombre total d'entrées : {len(df):,}")
    print(f"Nombre de joueurs uniques : {df['player_id'].nunique():,}")
    
    print("\nColonnes disponibles :")
    for i, col in enumerate(df.columns, 1):
        print(f"  {i:2d}. {col}")
    
    print("\nTypes de données :")
    print(df.dtypes)
    
    print("\nAperçu des premières lignes :")
    print(df.head())
    
    print("\nStatistiques descriptives :")
    print(df.describe())
    
    print("\nValeurs manquantes :")
    missing = df.isnull().sum()
    print(missing[missing > 0])
    
    print("\nInformations sur les blessures :")
    if 'injury_reason' in df.columns:
        print("Top 10 des types de blessures :")
        print(df['injury_reason'].value_counts().head(10))
else:
    print("Dataset 'player_injuries' non disponible")
```

## 13. Résumé global de tous les datasets

```python
print("\n" + "=" * 80)
print("RÉSUMÉ GLOBAL DE TOUS LES DATASETS")
print("=" * 80)

summary_data = []
total_rows = 0
total_columns = 0

for name, df in datasets.items():
    summary_data.append({
        'Dataset': name,
        'Lignes': f"{len(df):,}",
        'Colonnes': len(df.columns),
        'Taille_MB': f"{df.memory_usage(deep=True).sum() / 1024 / 1024:.1f}",
        'Valeurs_manquantes': f"{df.isnull().sum().sum():,}"
    })
    total_rows += len(df)
    total_columns += len(df.columns)

summary_df = pd.DataFrame(summary_data)
print(summary_df.to_string(index=False))

print(f"\nSTATISTIQUES GLOBALES :")
print(f"Nombre total de datasets : {len(datasets)}")
print(f"Nombre total de lignes : {total_rows:,}")
print(f"Nombre total de colonnes uniques : {total_columns}")
print(f"Taille totale en mémoire : {sum(df.memory_usage(deep=True).sum() for df in datasets.values()) / 1024 / 1024:.1f} MB")

print("\nANALYSE TERMINÉE - Tous les datasets ont été analysés !")
```
